## 4. Funciones / Bucles

Son estructuras para repetir un bloque de código múltiples veces, de manera controlada.

Hay dos tipos:
- `for`: itera sobre una secuencia (como una lista, una tupla, un diccionario o una cadena) o cualquier objeto iterable. Se define con for, seguida de una variable de iteración, la palabra clave in, y luego la secuencia o iterable sobre el que se va a iterar. REQUIERE DEFINIR PREVIAMENTE LA VARIABLE.

- `while`: ejecuta un bloque de código mientras se cumpla una condición específica. Se define utilizando la palabra clave while, seguida de la condición a evaluar. El bloque se ejecuta hasta que la condición deje de ser verdadera.



#### Ejemplo `for` usando `range`

````python
# Mostrar los cuadrados de los números del 1 al 5

for i in range(1, 6):
# for i in [1,2,3,4,5]:
    print(f"El cuadrado de {i} es {i**2}")
````

#### Ejemplo `for` usando `list`

````python
# Lista de palabras
palabras = ["autos", "jets", "aviones", "barcos"]

for p in palabras:
    if len(p) > 5:
        print(f"La palabra '{p}' tiene más de 4 letras")
    else:
        print(f"La palabra '{p}' tiene 4 letras o menos")
````

#### Ejemplo `while`

````python
# Sumar números naturales hasta que la suma pase de 50

suma = 0
numero = 1

while suma <= 50:
    # suma = suma + numero
    suma += numero
    print(f"Agrego {numero}, suma actual: {suma}")
    numero += 1

print("Se superó el límite de 50")
````

## 5. Introducción a Programación Orientada a Objetos



### ¿Por qué POO?

La **Programación Orientada a Objetos** nos permite:

1. **Agrupar datos y funciones** que operan sobre esos datos. Imaginemos que al tener múltiples funciones, quien tome el archivo no sabe qué objetos pueden introducirse como argumentos, si se pisan la definición entre archivos habrá problemas, etc.
2. **Reutilizar código** mediante herencia. Esto es asimilable a tener `child classes`, donde una clase inferior hereda de la superior ciertas funciones o definiciones.



3. **Entender librerías** como scikit-learn, pandas, numpy. Todas las librerías usan POO

### Conceptos clave

- **Clase**: plantilla o "molde" para crear objetos
- **Objeto (instancia)**: ejemplar concreto de una clase
- **Atributos**: variables que pertenecen a un objeto
- **Métodos**: funciones que pertenecen a un objeto
- **`self`**: referencia al objeto mismo dentro de la clase

### Primera clase: Alumno

Transformemos nuestro diccionario en una clase:

````python
class Alumno:
    """Clase para representar un alumno con sus datos y operaciones"""

    institucion = 'FIUBA' # Atributo de CLASE (Se guardan en la clase misma). COMPARTIDO

    def __init__(self, nombre, edad, notas):
        """Constructor: se ejecuta al crear una instancia"""
        self.nombre = nombre      # Atributo de INSTANCIA (Se guardan en cada objeto (no en la clase))
        self.edad = edad
        self.notas = notas

    def imprimir(self):
        print('hola')

    def promedio(self):
        """Método: calcula el promedio de notas"""
        return sum(self.notas) / len(self.notas)

    def agregar_nota(self, nota):
        """Método: agrega una nota nueva"""
        self.notas.append(nota)

    def __repr__(self):
        """Método especial: representa el objeto como string"""
        return f"Alumno(nombre='{self.nombre}', edad={self.edad}, notas={self.notas})"

````

````python
agustin = Alumno('Agustin',31,[6,8])
````

````python
agustin.imprimir()
````

````python
agustin.promedio()
````

````python
# Crear instancias (objetos)
catalina = Alumno('Catalina', 21, [8, 9, 7, 8])
sebastian = Alumno('Sebastián', 23, [9, 10, 8])

print(catalina)
print(sebastian)
````

````python
catalina.institucion
````

````python
# Acceder a atributos
print(f"Nombre: {catalina.nombre}")
print(f"Edad: {catalina.edad}")
print(f"Notas: {catalina.notas}")
````

````python
# Usar métodos
print(f"Promedio de {catalina.nombre}: {catalina.promedio()}")

# Agregar una nota
catalina.agregar_nota(10)
print(f"Notas actualizadas: {catalina.notas}")
print(f"Nuevo promedio: {catalina.promedio()}")
````

### Ejemplo con scikit-learn

Las librerías de Data Science usan POO extensivamente. Veamos un ejemplo con scikit-learn:

````python
from sklearn.linear_model import LinearRegression

# Crear instancia del modelo (como creamos un Alumno)
modelo = LinearRegression()

print("Estado ANTES del fit:")
print(modelo.__dict__)
print("\nAtributos disponibles:")
print([attr for attr in dir(modelo) if not attr.startswith('_')])
````

````python
# Datos de ejemplo: y = 2*x + 1
X = [[1], [2], [3], [4], [5]]
y = [3, 5, 7, 9, 11]

# Entrenar el modelo (método fit)
modelo.fit(X, y)

print("\nEstado DESPUÉS del fit:")
print(modelo.__dict__)
````

````python
# Atributos aprendidos (terminan en _)
print(f"Coeficiente (pendiente): {modelo.coef_}")
print(f"Intercepto: {modelo.intercept_}")

# Predicción (método predict)
nuevos_x = [[6], [7]]
predicciones = modelo.predict(nuevos_x)
print(f"\nPredicciones para {nuevos_x}: {predicciones}")
````
